```python
import pandas as pd
import sqlite3


df = pd.read_csv('g32_Farmers_Markets.csv', sep=';')


conn = sqlite3.connect('data/trabalhopc_project.db')

# TABELA FARMERS
df_farmers = df[['farmer_id', 'name', 'creation_date']].drop_duplicates()
df_farmers.columns = ['id', 'name', 'creation_date']
df_farmers.to_sql('farmers', conn, if_exists='append', index=False)

# TABELA MARKETS
df_markets = df[['market_id', 'title', 'category']].drop_duplicates()
df_markets.columns = ['id', 'title', 'category']
df_markets.to_sql('markets', conn, if_exists='append', index=False)

# TABELA WORKERS
df_workers = df[['worker_id', 'extra_info', 'farmer_id']].dropna(subset=['worker_id']).drop_duplicates()
df_workers.columns = ['id', 'extra_info', 'farmer_id']
df_workers.to_sql('workers', conn, if_exists='append', index=False)

# TABELA TRANSACTIONS
df_trans = df[['transaction_date', 'amount', 'farmer_id', 'market_id']]
df_trans.to_sql('transactions', conn, if_exists='append', index=False)

conn.close()

print("Os dados foram inseridos na base de dados.")
```

```python
import pandas as pd
import sqlite3

df = pd.read_csv('g32_Farmers_Markets.csv', sep=';')
conn = sqlite3.connect('data/trabalhopc_project.db')

df_farmers = df[['farmer_id', 'name', 'creation_date']].drop_duplicates()
df_farmers.columns = ['id', 'name', 'creation_date']
df_farmers.to_sql('farmers', conn, if_exists='replace', index=False)

df_markets = df[['market_id', 'title', 'category']].drop_duplicates()
df_markets.columns = ['id', 'title', 'category']
df_markets.to_sql('markets', conn, if_exists='replace', index=False)

df_workers = df[['worker_id', 'extra_info', 'farmer_id']].dropna(subset=['worker_id']).drop_duplicates()
df_workers.columns = ['id', 'extra_info', 'farmer_id']
df_workers.to_sql('workers', conn, if_exists='replace', index=False)

df_trans = df[['transaction_date', 'amount', 'farmer_id', 'market_id']]
df_trans.to_sql('transactions', conn, if_exists='replace', index=False)

conn.close()
print("Os dados foram inseridos na base de dados.")
```
