```python
import pandas as pd
from sqlalchemy import create_engine

```

```python

engine = create_engine('sqlite:///data/trabalhopc_project.db')


df = pd.read_csv("g32_Farmers_Markets_Final.csv", sep=';')
```

```python
# A. Tabela markets
df_markets = df[['market_id', 'title', 'category']].drop_duplicates(subset='market_id')
df_markets.rename(columns={'market_id': 'id'}, inplace=True)

# B. Tabela MarketType
df_market_type = pd.DataFrame()
df_market_type['id_market'] = df_markets['id']

df_market_type['market_type'] = df_market_type['id_market'].apply(
    lambda x: "Mercado de Fruta" if x % 3 == 1 else ("Mercado de Legumes" if x % 3 == 2 else "Mercado de Flores")
)

# C. Tabela farmers 
df_farmers = df[['farmer_id', 'name', 'creation_date']].drop_duplicates(subset='farmer_id')
df_farmers.rename(columns={'farmer_id': 'id'}, inplace=True)

# D. Tabela workers 
df_workers = df[['worker_id', 'extra_info', 'farmer_id']].drop_duplicates(subset='worker_id')
df_workers.rename(columns={'worker_id': 'id'}, inplace=True)

# E. Tabela Transactions
df_transactions = df[['farmer_id', 'market_id', 'transaction_date', 'amount']].copy()

# 1. Criar a coluna 'id' automática (começando em 1)
df_transactions.insert(0, 'id', range(1, len(df_transactions) + 1))

# 2. Gravar com o nome 'Transactions'
df_transactions.to_sql('Transactions', con=engine, if_exists='replace', index=False)

```

```python
df_markets.to_sql('Market', con=engine, if_exists='replace', index=False)
df_market_type.to_sql('MarketType', con=engine, if_exists='replace', index=False)
df_farmers.to_sql('Farmer', con=engine, if_exists='replace', index=False)
df_workers.to_sql('Worker', con=engine, if_exists='replace', index=False)
df_transactions.to_sql('Transactions', con=engine, if_exists='replace', index=False)

# Teste da tabela Market_Type
print("Dados carregados na tabela MarketType:")
result = pd.read_sql('MarketType', con=engine)
display(result.head(5))

# Fechar a ligação
engine.dispose()
```
